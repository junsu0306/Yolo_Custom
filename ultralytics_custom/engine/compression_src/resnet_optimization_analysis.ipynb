```python
import torch
from models.resnet import ResNet18
from pruning.resnet import resnet18_pruning
from reducing.resnet import resnet18_reduce
import copy
```

```python
model = ResNet18()
```

```python
model
```

```python
# 프루닝 전 모델 가중치 저장
pre_pruning_weights = copy.deepcopy(model.state_dict())
```

```python
# 프루닝 수행 (sparsity를 0.5로 설정: 50% 필터를 제거)
sparsity = 0.5
resnet18_pruning(model, sparsity)
```

```python
# 프루닝 후 모델 가중치 저장
post_pruning_weights = model.state_dict()
```

```python
# 프루닝 전후의 가중치 비교
for layer_name, pre_weight in pre_pruning_weights.items():
    post_weight = post_pruning_weights[layer_name]
    if torch.equal(pre_weight, post_weight):
        pass
        # print(f"{layer_name}는 프루닝되지 않았습니다.")
    else:
        diff = torch.sum(pre_weight != post_weight).item()
        print(f"{layer_name}의 가중치 중 {diff}개의 요소가 프루닝되었습니다.")
```

```python
# 새로운 모델 생성
reduced_model = ResNet18()
```

```python
# 모델 크기 줄이기
resnet18_reduce(model, reduced_model)
```
