```python
import torch 

model = torch.load("./reduced_model_50_book.pth")


total_zero = 0
total_params = sum(p.numel() for p in model.parameters())
with torch.no_grad():
    for p in model.parameters():
        # 정확히 0인 요소 개수
        z = (p == 0).sum().item() 
        total_zero += z

print(f"Before reducing: {21376751}")
print(f"Total parameters: {total_params}")
print(f"zero parameters: {total_zero}")



```
